```python
import pandas as pd
ruta = "../data/raw/INM_estatal_dic25.csv"
df = pd.read_csv(ruta, dtype={"clave_ent": str})
df.head()
```

```python
df = df[df["anio"].isin([2024, 2025])].copy()
df.shape
```

```python
meses = {"Enero": 1, "Febrero": 2, "Marzo": 3, "Abril": 4, "Mayo": 5, "Junio": 6,
         "Julio": 7, "Agosto": 8, "Septiembre": 9, "Octubre": 10, "Noviembre": 11, "Diciembre": 12}     # completa los 12
df["mes_num"] = df["mes"].map(meses)
df["mes_num"].isna().sum()   # debe dar 0; si no, escribiste mal algún mes
```

```python
df["fecha"] = pd.to_datetime(df["fecha"])
df.dtypes   # fecha debe decir datetime64
```

```python
df = df.rename(columns={
    "clave_ent": "clave_entidad",
    "entidad_federativa": "estado",
    "entidad": "estado_nombre_oficial",
    "bien_juridico_afectado": "bien_juridico",
    "incidencia_delictiva": "incidencia",
})
df.columns
```

```python
df.to_csv("../data/processed/fact_incidencia_mensual.csv", index=False, encoding="utf-8-sig")
```

```python
dim_estado = df[["clave_entidad", "estado", "estado_nombre_oficial"]].drop_duplicates()
dim_estado.to_csv("../data/processed/dim_estado.csv", index=False, encoding="utf-8-sig")
dim_delito = df[["bien_juridico", "tipo_delito"]].drop_duplicates()     # bien jurídico y tipo de delito
dim_delito.to_csv("../data/processed/____.csv", index=False, encoding="utf-8-sig")
len(dim_estado), len(dim_delito)
```
